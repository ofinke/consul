```python
from consul.core.schemas import ToolConfig
from consul.db.handler import get_db_handler
from consul.db.tables import AppConfigTable
from consul.tools.registry import ToolsRegistry, get_tool_registry
from sqlalchemy import select
```

```python
def get_all_tools_registry() -> ToolsRegistry:
    """Return instance of ToolRegistry with all tools from all servers."""
    handler = get_db_handler()
    mcp_configs = handler.load(
        AppConfigTable, statement=select(AppConfigTable.name).where(AppConfigTable.validation_model == "MCPConfig")
    )
    config = ToolConfig(servers=[config[0] for config in mcp_configs])
    return get_tool_registry(config)
```

```python
all_tools_registry = get_all_tools_registry()
await all_tools_registry.register_tools()
```
